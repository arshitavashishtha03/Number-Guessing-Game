```python
import time
import math
import random

a = input("my name is BOT, what's your name???: ")
print()

print("HI", a, "I chose a number betweeen 1 to 100:")
print()

# Pause for 2 seconds
time.sleep(2)

for i in range(1, 101):
    print(i)
    time.sleep(0.05)

print()
print("now, you have to Guess what number I chose,")
time.sleep(1)
print("I'll be giving you hints for that !!!!! and You will get only 7 attempts ")
time.sleep(1)
print()

ready = input("Are you ready? (yes/no): ").strip().lower()
print()

if ready == "yes":
    print("Awesome! Let's start the game.")
    N = random.randint(1, 100)
    attempts = 0

    while attempts < 7:
        try:
            guess = int(input("Make a random guess: "))
            attempts += 1

            if guess < N:
                print("The secret number is higher than your guess. Try again!")
                print()
            elif guess > N:
                print("The secret number is lower than your guess. Try again!")
                print()
            else:
                print("Congratulations! You guessed the correct number!")
                break
        except ValueError:
            print("Please enter a valid integer!")
    else:
        print("You have used all 7 attempts! the number was :" , N)

else:
    print("No problem! Let me know when you are ready to play.")
print()
```
